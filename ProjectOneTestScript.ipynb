```python
# Python Testing Script

from CRUD_Python_Module import AnimalShelter
from pymongo.errors import PyMongoError

# Instantiate the AnimalShelter class with correct credentials
try:
    shelter = AnimalShelter()
    print("Connected to MongoDB successfully.")
except PyMongoError as e:
    print(f"Could not connect to MongoDB: {e}")
```

```python
# Test creating a new animal record
new_animal = {
    "name": "TestDog", 
    "animal_type": "Dog",
    "breed": "Dachshund",
    "age_upon_outcome": "2 years"
    }

try:
    success = shelter.create(new_animal)
    if success:
        print("Record created successfully")
    else:
        print("Record was not created.")
except PyMongoError as e:
    print(f"Error creating record: {e}")
```

```python
# Test reading all animal records
try:
    animals = shelter.read()
    print(f"Retrieved {len(animals)} records from MongoDB")
except PyMongoError as e:
    print(f"Error reading records: {e}")
```

```python
# Test reading new document
try:
    dogs = shelter.read({"name": "TestDog"})
    print(dogs)
except PyMongoError as e:
    print(f"Error reading records: {e}")
```

```python
# Test update function

# Verify the record exists
print("Before update:")
dogs = shelter.read({"name": "TestDog"})
print(dogs)

# Update the record (change age and breed)
update_count = shelter.update(
    {"name": "TestDog"},  # query
    {"age_upon_outcome": "2 years",
     "breed": "Golden Retriever"}  # new values
)
print(f"Documents updated: {update_count}")

# Verify the update
print("After update:")
dogs = shelter.read({"name": "TestDog"})
print(dogs)
```

```python
# Test the delete function
delete_count = shelter.delete({"name": "TestDog"})
print(f"Documents deleted: {delete_count}")

# Verify deletion
print("After delete:")
dogs = shelter.read({"name": "TestDog"})
print(dogs)
```
